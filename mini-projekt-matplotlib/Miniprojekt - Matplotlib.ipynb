```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(2026)
area = rng.uniform(30, 180, 250)
distance = rng.uniform(0, 20, 250)
price = (
    125 + 3.2 * area - 2.5 * distance
    + rng.normal(0, 45, 250)
)
price = np.clip(price, 70, None)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 7))

# Histogram
axes[0, 0].hist(
    price, bins=25,
    color="steelblue",
    edgecolor="white"
)
axes[0, 0].set_title("Rozkład cen mieszkań")

# Cena a powierzchnia
axes[0, 1].scatter(
    area, price,
    alpha=0.5, color="teal"
)
axes[0, 1].set_title("Cena a powierzchnia")

# Cena a odległość
axes[1, 0].scatter(
    distance, price,
    alpha=0.5, color="darkorange"
)
axes[1, 0].set_title("Cena a odległość od centrum")

# Podział mieszkań według powierzchni
small = price[area < 70]
medium = price[(area >= 70) & (area < 120)]
large = price[area >= 120]

axes[1, 1].boxplot(
    [small, medium, large],
    labels=["Małe", "Średnie", "Duże"]
)
axes[1, 1].set_title("Cena według wielkości mieszkania")

for ax in axes.flat:
    ax.grid(alpha=0.2)
    ax.set_axisbelow(True)
fig.suptitle("Eksploracyjna analiza cen mieszkań")
fig.tight_layout()

plt.show()
```
